<span style="font-size:80%">[문제1]<br><br></span>
두 수를 받은 뒤 곱한 결과를 반환하는 oz 함수를 만들어주세요


<span style="color:red">답변</span>

````python
def oz(x, y):
    return x * y

try:
    x = int(input("첫번째 수: "))
    y = int(input("두번째 수: "))
    print(f'두 수의 곱은 {oz(x, y)}입니다.')
except ValueError:
    print("정수만 입력해주세요!")
````

### Chapter18 : `Python의 Class 개념에 대해 복습하고 문제풀기`

<span style="font-size:80%">[문제1]<br><br></span>
```클래스 greeting 메서드를 호출하기 위해 필요한 코드를 작성해주세요```

class Person:<br>
&emsp;def greeting(self):<br>
&emsp;&emsp;print('hello')<br>

<span style="color:red">답변</span>

````python
class Person:
# Person 이라는 클래스 정의
    def greeting(self):
    # class 내부에서 정의된 함수, 즉 이 함수에서 작동하는 매서드
    # 매서드는 반드시 첫번째 매개변수로 self 를 써야한다.
        print('hello')

person = Person()
# num = int() 와 똑같이 실제 Person이라는 클래스를 가진 데이터가 들어갈 자리가 생김
person.greeting()
# greeting() 으로 비워서 호출했지만 파이썬에서 자동으로 첫번째 매개변수를
# (self)로 넣어 person.greeting(person) 이 실행된다.
````
