[문제 6] 0부터 시작하여 n까지의 숫자 중에서 홀수만 생성하는 제너레이터 odd_numbers를 작성하세요.

```python
def odd_numbers(n: int):
    for i in range(n+1) :
    # 마지막 숫자인 n을 포함하려면 n+1 까지 돌려야함
        if i % 2 != 0 :
            yield i
            #print(i)가 아니라 제너레이터를 만들어야함
            #대신 홀수 한개를 잡아내면 멈춰버림

input_num = odd_numbers(int(input("원하는 숫자를 입력하세요 :")))
#input_num 함수에 odd_nubers 제너레이터를 넣는다.

print(next(input_num))
# 위의 input으로 15를 넣었다면 input_num는 odd_numbers(15)
# 그러나 odd_numbers 제너레이터는 부를 때 마다 새로 생성되므로
# input_num 안에서 한개의 제너레이터를 생성한 뒤에 해당 제너레이터를 계속 사용해야함

print(next(input_num))
print(next(input_num))
print(next(input_num))
print(next(input_num))
print(next(input_num))
# 이렇게 반복하면 되지만, 더이상 들어오는 값이 없을 때 까지 반복해야 하므로
# for 문을 사용하면 된다.
```

```python
def odd_numbers(n: int):
    for i in range(n+1) :
        if i % 2 != 0 :
            yield i

input_num = int(input("원하는 숫자를 입력하세요 :"))
for num in odd_numbers(input_num):
    print(num)
```

[문제 10] 팩토리얼을 계산하는 재귀 함수 factorial을 작성하세요.

```python
def factorial(n):
    if n == 0 :
        return 1
    else :
        return n * (factorial(n-1))

factorial_input = int(input("숫자를 입력하세요 :"))
print(factorial(factorial_input))
```

[문제 11] 문자열 리스트를 받아, 각 문자열의 길이를 출력하는 함수 print_lengths를 작성하세요.

```python
def print_lengths(str_list):
        result = []
        for i in str_list:
                result.append(len(i))
        return result

str_list_input = input("문자열들을 공백으로 나눠 입력하세요: ").split()

print(print_lengths(str_list_input))

#리스트로 받아서 print_lengths에 리스트를 통째로 넣기 때문에 *를 사용하지 않아도 된다.
```
