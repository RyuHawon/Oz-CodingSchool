__[문제3]__

숫자를 건넬 때 일부 자릿수를 영단어로 바꾼 카드를 건네주면 프로도는 원래 숫자를 찾는 게임입니다.

다음은 숫자의 일부 자릿수를 영단어로 바꾸는 예시입니다.

1478 → "one4seveneight"  
234567 → "23four5six7"  
10203 → "1zerotwozero3"    

이렇게 숫자의 일부 자릿수가 영단어로 바뀌어졌거나,   
혹은 바뀌지 않고 그대로인 문자열 s가 주어졌을때
s가 의미하는 원래 숫자를 출력하도록 코드를 작성해주세요

s = "77three4one

---
참고로 각 숫자에 대응되는 영단어는 다음 표와 같습니다.

숫자	영단어  
0	zero  
1	one  
2	two  
3	three  
4	four  
5	five  
6	six  
7	seven  
8	eight

```python
number = {
    0: "zero",
    1: "one",
    2: "two",
    3: "three",
    4: "four",
    5: "five",
    6: "six",
    7: "seven",
    8: "eight",
    9: "nine"
}
s = "77three4one"

for num, eng in number.items():
    s = s.replace(eng, str(num))

# for문 한번에 키와 밸류 두가지 값을 모두 사용 할 수 있도록 .item 매서드 사용
# replace 매서드는 (지울내용, 대체내용)

print(s)
```

__[문제4]__

행렬의 덧셈은 행과 열의 크기가 같은 두 행렬의 같은 행, 같은 열의 값을 서로 더한 결과가 됩니다. 

 행렬 arr1과 arr2가 주어졌을 때, 행렬 덧셈의 결과를 출력하는 코드를 작성해주세요(result와 같은 결과가 출력되도록 작성)

arr1 = [[1,2],[2,3]]

arr2 = [[3,4],[5,6]]

result = [[4,6],[7,9]]

```python
arr1 = [[1,2],[2,3]]
arr2 = [[3,4],[5,6]]
sum_arr = []

for x, y in zip(arr1, arr2) :
#[1,2]는 x, [3,4]는 y로 불러온다.
    box_arr = []
    for i, j in zip(x, y):
    # i 에는 1,3을, j에는 2,4를 불러온 후 i+j 값을 sum_arr에 저장해준다.
        box_arr.append(i + j)
    sum_arr.append(box_arr)

print(sum_arr)
```

__[문제5]__

양의 정수 x가 하샤드 수이려면 x의 자릿수의 합으로 x가 나누어져야 합니다.   
예를 들어 18의 자릿수 합은 1+8=9이고, 18은 9로 나누어 떨어지므로 18은 하샤드 수입니다.   
주어진 x를 이용해 x가 하샤드 수인지 아닌지 검사하는 코드를 작성해주세요

x = 17

----

입출력 예시
x	결과값  
10	true  
12	true  
11	false  
13	false  

```python
x = (input())
x_sum = 0

for i in range(len(x)) :
    # 제발 for 문에 len(x)라고 써서 정수 하나만 넣는 짓을 그만하자.
    # range()를 꼭 씌우자.
    x_sum += int(x[i])
    #result_x = x_sum + int(x[i]) <- for문 돌 때 마다 원래 있던 값 지우고 새 값넣음...
    #누적시키려면 변수 한개로 += 연산자를 

if int(x) % x_sum == 0 :
    print(f'{x}는 하샤드 수가 맞습니다.')
else :
    print(f'{x}는 하샤드 수가 아닙니다.')
```

__[문제6]__

String형 배열 seoul의 element중 "Kim"의 인덱스 x를 찾아, "김서방은 x에 있다"는 String을 출력하는 코드를 작성해주세요.   

seoul에 "Kim"은 오직 한 번만 나타나며 잘못된 값이 입력되는 경우는 없습니다.

seoul = ["Jane", "Kim"]

<br>
출력 예) "김서방은 1에 있다"

```python
seoul = ["Jane", "Kim"]

for i in range(len(seoul)) :
    if seoul[i] == "Kim" :
        print(f'김서방은 {i}에 있다.')
        break

# print(f'김서방은 {seoul.index("Kim")}에 있다.') 매서드를 사용하면 더 쉬웠을것
```
