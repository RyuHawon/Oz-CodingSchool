[문제 3]

사용자로부터 여러 개의 숫자를 입력받아 리스트에 저장한 후, 사용자가 'done'이라고 입력할 때까지 계속 입력받으세요. 

- 입력 중 숫자가 아닌 값이 입력되면 ValueError를 처리
- 숫자 외 값이 입력된 경우 어떤 값이 잘못 입력되었는지 사용자에게 알려주는 문구 출력<br>
  (예: f"잘못된 값: {변수}. 숫자를 입력해주세요.")



````python
num_list = []
# while 문 안에 리스트를 만들면 한바퀴 돌 때 마다 기존의 내용물을 버리고 새 리스트를 생성

while True:
    try:
        x = input('저장할 숫자를 입력하세요. 입력 종료 : done입력')
        if x == 'done':
            break
        num_list.append(int(x))
    except ValueError:
        print(f'숫자만 입력해주세요. 현재 입력 값: {x}')

print(f'저장된 숫자 : {num_list}')
````

[문제 4]

사용자 정의 예외 NegativeNumberError를 생성하고, 사용자로부터 숫자를 입력받아 음수일 경우 NegativeNumberError를 발생시키는 프로그램을 작성하세요. 

- 예외 처리 구문을 사용하여 음수 입력 시 "양수를 입력해주세요."라는 메시지를 출력

````python
#이해가 잘 안돼서 정답지 따라치면서 한번 이해해봄

class NegativeNumberError(Exception):
# Exception 클래스에서 파생
    def __init__(self, message) -> None:
    # 에러가 생성될 때 message(에러안내메세지)를 실행시키게 된다.
        self.contact_name = "NegativeNumberError"
        # 이름 지정
        self.message = message
        # 에러 메세지 이름도 저장
        super().__init__(message)
        # 부모클래스 Exception에도 메세지 전달
    def __str__(self) -> str:
        return f"{self.contact_name} : {self.message}"
    
try:
    number = int(input("양수를 입력해주세요 :"))
    if number < 0:
        raise NegativeNumberError("양수만 입력해주세요!")
        # NegativeNumberError 라는 에러를 일으켜라.
        # 에러가 생성되면 message(에러안내메세지) 넣어주기로 자리를 맡아놨음.
        # __init__(self, message) message자리에 "양수만 입력해주세요!"가 들어감
except NegativeNumberError as e:
    print(e)
    # NegativeNumberError를 실행해야하는데, 어떤걸 print 해야해?
    # 파이썬은 자동으로 __str__ 함수를 찾아서 실행시키게 설계됐음 (이름이 __str__ 이어야함)
    # 굳이 print(e.__str__())라고 쓰지 않아도 된다.
````

````python
# 그냥 내 생각대로 만들어봄

class NegativeNumberError(Exception):
    def __init__(self, message="양수만 입력해주세요!"):
        # 음수숫자에러이므로 기본적인 메세지를 저장해둠
        self.name = "NegativeNumberError"
        self.message = message
        # return f"{self.name} : {self.message}"
        # __init__은 리턴을 가질 수 없다.
        super().__init__(message)
        # 일단 부모(Exception)이 가진 기능들 다 계승해
    def __str__(self):
    #__str__ 은 내가 다시 쓸거야. 오버라이딩.
        return f"{self.name} : {self.message}"
        # __str__ 은 리턴값을 가질 수 있다.

try:
    number = int(input("양수를 입력해주세요 :"))
    if number < 0:
        raise NegativeNumberError()
        # 따로 넣을 내용이 없으므로 기본 메세지가 출력되게 비워둔다.
except NegativeNumberError as e:
    print(e)
````

[문제 7]

함수 실행 시간을 측정하는 데코레이터 timer_decorator를 작성하세요. 

임의의 함수가 실행되는 시간을 측정하여 "함수 실행 시간: x.xx초" 형태로 출력하세요.

````python
import time

def timer_decorator(funcion):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        # 함수 실행 전 시간 저장
        result = funcion(*args, **kwargs)
        # 받은 내용 그대로 원본 함수에 넣어서 일단 함수는 정상적으로 작동되게 한다.
        # 인자 *args, **kwargs 는 뭐든지 다 받아서 전달할 수 있는 주머니
        end_time = time.time()
        # 함수 실행 후 시간 저장
        run_time = end_time - start_time
        print(f'함수 실행 시간: {run_time:.2f}초')
        # .2f사용으로 무조건 소수점 두번째 자리에서 끝나도록한다.
        return result
    return wrapper
````
